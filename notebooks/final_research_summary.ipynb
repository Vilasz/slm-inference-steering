# Final Research Summary

Este notebook consolida os artefatos das fases 1 a 8 para apresentacao. Ele le tabelas e figuras geradas por `scripts/generate_final_report_artifacts.py`.

A primeira celula carrega caminhos padrao e mostra se o resumo final ja existe.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

ROOT = Path('..').resolve()
SUMMARY_PATH = ROOT / 'runs' / 'final_summary.json'
TABLES = ROOT / 'reports' / 'tables'
FIGURES = ROOT / 'reports' / 'figures'
SUMMARY_PATH.exists()
```

Carregamos o resumo em JSON. Se ele nao existir, rode o script de artefatos finais.

```python
if SUMMARY_PATH.exists():
    final_summary = json.loads(SUMMARY_PATH.read_text(encoding='utf-8'))
else:
    final_summary = {'notes': ['Execute scripts/generate_final_report_artifacts.py']}
final_summary
```

A tabela principal reune custo, acuracia e metadados das runs encontradas.

```python
main_path = TABLES / 'main_results.csv'
main = pd.read_csv(main_path) if main_path.exists() else pd.DataFrame()
main.head(20)
```

As figuras abaixo sao geradas automaticamente. Elas viram material direto para slides: fronteira custo-acuracia e cobertura experimental.

```python
for name in ['accuracy_cost_scatter.png', 'phase_coverage.png']:
    path = FIGURES / name
    if path.exists():
        display(Image(filename=str(path)))
    else:
        print(f'Figura ausente: {name}')
```

Use esta celula para separar evidencias por fase: baseline, probing, steering causal, politicas adaptativas e robustez.

```python
if main.empty or 'phase' not in main:
    phase_summary = pd.DataFrame()
else:
    phase_summary = main.groupby('phase').agg(
        runs=('run_label', 'count'),
        best_accuracy=('observed_best_of_n', 'max'),
        best_pass_at_1=('strict_pass_at_1', 'max'),
        min_tokens=('mean_tokens_until_success_or_budget', 'min'),
    ).reset_index()
phase_summary
```

Narrativa sugerida para apresentacao: primeiro mostre a fronteira do baseline, depois a separabilidade latente, em seguida o teste causal com controles, e por fim a politica adaptativa como resultado de eficiencia.
